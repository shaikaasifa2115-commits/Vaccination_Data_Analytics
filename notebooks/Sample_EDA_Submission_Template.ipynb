## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Data Wrangling Code

# Create copies so the original loaded datasets are preserved
coverage_clean = coverage.copy()
incidence_rate_clean = incidence_rate.copy()
reported_cases_clean = reported_cases.copy()
vaccine_introduction_clean = vaccine_introduction.copy()
vaccine_schedule_clean = vaccine_schedule.copy()

# ---------------------------------------------------------
# 1. Remove duplicate rows
# ---------------------------------------------------------
for df in [
    coverage_clean,
    incidence_rate_clean,
    reported_cases_clean,
    vaccine_introduction_clean,
    vaccine_schedule_clean
]:
    df.drop_duplicates(inplace=True)

# ---------------------------------------------------------
# 2. Clean text columns
# ---------------------------------------------------------
for df in [
    coverage_clean,
    incidence_rate_clean,
    reported_cases_clean,
    vaccine_introduction_clean,
    vaccine_schedule_clean
]:
    text_columns = df.select_dtypes(include="object").columns
    
    for col in text_columns:
        df[col] = df[col].str.strip()

# ---------------------------------------------------------
# 3. Convert YEAR to numeric and remove invalid YEAR rows
# ---------------------------------------------------------
cleaned_datasets = {
    "Coverage": coverage_clean,
    "Incidence Rate": incidence_rate_clean,
    "Reported Cases": reported_cases_clean,
    "Vaccine Introduction": vaccine_introduction_clean,
    "Vaccine Schedule": vaccine_schedule_clean
}

for name, df in cleaned_datasets.items():
    df["YEAR"] = pd.to_numeric(df["YEAR"], errors="coerce")
    df.dropna(subset=["YEAR"], inplace=True)
    df["YEAR"] = df["YEAR"].astype(int)

# ---------------------------------------------------------
# 4. Clean Coverage numerical fields
# ---------------------------------------------------------
coverage_clean["TARGET_NUMBER"] = pd.to_numeric(
    coverage_clean["TARGET_NUMBER"], errors="coerce"
)

coverage_clean["DOSES"] = pd.to_numeric(
    coverage_clean["DOSES"], errors="coerce"
)

coverage_clean["COVERAGE"] = pd.to_numeric(
    coverage_clean["COVERAGE"], errors="coerce"
)

# Invalid negative doses
coverage_clean.loc[
    coverage_clean["DOSES"] < 0, "DOSES"
] = np.nan

# Coverage must be between 0 and 100
coverage_clean.loc[
    (coverage_clean["COVERAGE"] < 0) |
    (coverage_clean["COVERAGE"] > 100),
    "COVERAGE"
] = np.nan

# ---------------------------------------------------------
# 5. Clean Incidence Rate
# ---------------------------------------------------------
incidence_rate_clean["DENOMINATOR"] = pd.to_numeric(
    incidence_rate_clean["DENOMINATOR"], errors="coerce"
)

incidence_rate_clean["INCIDENCE_RATE"] = pd.to_numeric(
    incidence_rate_clean["INCIDENCE_RATE"], errors="coerce"
)

incidence_rate_clean.loc[
    incidence_rate_clean["DENOMINATOR"] < 0,
    "DENOMINATOR"
] = np.nan

incidence_rate_clean.loc[
    incidence_rate_clean["INCIDENCE_RATE"] < 0,
    "INCIDENCE_RATE"
] = np.nan

# ---------------------------------------------------------
# 6. Clean Reported Cases
# ---------------------------------------------------------
reported_cases_clean["CASES"] = pd.to_numeric(
    reported_cases_clean["CASES"], errors="coerce"
)

reported_cases_clean.loc[
    reported_cases_clean["CASES"] < 0,
    "CASES"
] = np.nan

# ---------------------------------------------------------
# 7. Clean Vaccine Schedule numerical field
# ---------------------------------------------------------
vaccine_schedule_clean["SCHEDULEROUNDS"] = pd.to_numeric(
    vaccine_schedule_clean["SCHEDULEROUNDS"], errors="coerce"
)

vaccine_schedule_clean.loc[
    vaccine_schedule_clean["SCHEDULEROUNDS"] < 0,
    "SCHEDULEROUNDS"
] = np.nan

# ---------------------------------------------------------
# 8. Display cleaned dataset sizes
# ---------------------------------------------------------
print("Data wrangling completed successfully!\n")

for name, df in cleaned_datasets.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
````

### What all manipulations have you done and insights you found?

### Data Wrangling Manipulations and Insights

The following data-wrangling operations were performed:

1. Removed duplicate rows from all five datasets.
2. Removed leading and trailing spaces from text columns.
3. Converted YEAR into numeric format and removed records with invalid or missing years.
4. Converted numerical fields into appropriate numeric data types.
5. In the Coverage dataset, negative DOSES values were treated as invalid.
6. Coverage values outside the valid 0–100% range were treated as invalid.
7. In the Incidence Rate dataset, negative DENOMINATOR and INCIDENCE_RATE values were treated as invalid.
8. In the Reported Cases dataset, negative CASES values were treated as invalid.
9. In the Vaccine Schedule dataset, negative SCHEDULEROUNDS values were treated as invalid.
10. Missing health-related numerical observations were not blindly imputed because missing values may represent unavailable or unreported observations in the source data.

### Insights

The datasets contain a large number of records covering vaccination coverage, disease incidence, reported cases, vaccine introduction, and vaccination schedules across countries and years.

After wrangling, the datasets were reduced by one invalid/duplicate record where applicable, while the overall structure and information were preserved. The cleaned datasets are ready for further exploratory analysis and visualization.

#### Chart - 5

````python
# Chart - 5: Vaccination Coverage vs Incidence Rate

import matplotlib.pyplot as plt

# Prepare average vaccination coverage by country and year
coverage_avg = (
    coverage.groupby(["CODE", "YEAR"])["COVERAGE"]
    .mean()
    .reset_index()
)

# Prepare average incidence rate by country and year
incidence_avg = (
    incidence_rate.groupby(["CODE", "YEAR"])["INCIDENCE_RATE"]
    .mean()
    .reset_index()
)

# Merge both datasets
comparison = pd.merge(
    coverage_avg,
    incidence_avg,
    on=["CODE", "YEAR"],
    how="inner"
)

# Remove missing values
comparison = comparison.dropna(
    subset=["COVERAGE", "INCIDENCE_RATE"]
)

# Create scatter plot
plt.figure(figsize=(12, 7))

plt.scatter(
    comparison["COVERAGE"],
    comparison["INCIDENCE_RATE"],
    alpha=0.4
)

plt.title("Vaccination Coverage vs Incidence Rate")
plt.xlabel("Average Vaccination Coverage (%)")
plt.ylabel("Average Incidence Rate")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

#### Chart - 15 - Pair Plot

````python
# Chart 15 - Pair Plot

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Select required numerical columns
pair_data = coverage[["YEAR", "TARGET_NUMBER", "DOSES", "COVERAGE"]].copy()

# Convert columns to numeric
for col in ["YEAR", "TARGET_NUMBER", "DOSES", "COVERAGE"]:
    pair_data[col] = pd.to_numeric(pair_data[col], errors="coerce")

# Remove missing values
pair_data = pair_data.dropna()

# Take a random sample to make the plot faster
pair_data = pair_data.sample(
    n=min(2000, len(pair_data)),
    random_state=42
)

# Create pair plot
g = sns.pairplot(
    pair_data,
    diag_kind="hist"
)

g.fig.suptitle(
    "Pair Plot of Vaccination Coverage Variables",
    fontsize=18,
    fontweight="bold",
    y=1.02
)

plt.show()

````

##### 1. Why did you pick the specific chart?

A pair plot was selected to visualize the pairwise relationships and distributions among the numerical vaccination variables. It helps identify patterns, clusters, trends, and potential outliers across multiple variables simultaneously.

##### 2. What is/are the insight(s) found from the chart?

The pair plot shows that TARGET_NUMBER and DOSES have a noticeable positive relationship, with higher target numbers generally associated with higher doses. YEAR shows a broad distribution across the variables, while COVERAGE has a wider spread and several high-value observations. The plots also show some outliers and considerable variation in the data.
